```python
import pandas as pd
from pathlib import Path
from testgen.analysis_utils import *
from functools import reduce
```

```python
labels = get_sensors_labels()
labels
```

```python
# read results file
base_path = Path().cwd()
results_path = base_path.parent / "results"
    
gpt_results = get_best_results(results_path, "single_gpt*n-1*.json")[0]
llama3b_results = get_best_results(results_path, "single_llama-3.2-3*n-1*.json")[0]
llama90b_results = get_best_results(results_path, "single_llama-3.2-90*n-1*.json")[0]
mixtral_results = get_best_results(results_path, "single_mixtral*n-1*.json")[0]
```

```python
gpt_df, _ = result_to_df(labels, gpt_results, results_path)
llama3b_df, _ = result_to_df(labels, llama3b_results, results_path)
llama90b_df, _ = result_to_df(labels, llama90b_results, results_path)
mixtral_df, _ = result_to_df(labels, mixtral_results, results_path)
```

```python
gpt_df = gpt_df[["requirement", "accuracy"]]
llama3b_df = llama3b_df[["requirement", "accuracy"]]
llama90b_df = llama90b_df[["requirement", "accuracy"]]
mixtral_df = mixtral_df[["requirement", "accuracy"]]
```

```python
gpt_df.rename(
    columns={"accuracy": gpt_results.stem[: gpt_results.stem.find("_acc")]},
    inplace=True,
)
llama3b_df.rename(
    columns={"accuracy": llama3b_results.stem[: llama3b_results.stem.find("_acc")]},
    inplace=True,
)
llama90b_df.rename(
    columns={"accuracy": llama90b_results.stem[: llama90b_results.stem.find("_acc")]},
    inplace=True,
)
mixtral_df.rename(
    columns={"accuracy": mixtral_results.stem[: mixtral_results.stem.find("_acc")]},
    inplace=True,
)
```

```python
df = reduce(
    lambda left, right: pd.merge(left, right, on=["idx", "requirement"], how="inner"),
    [gpt_df, llama3b_df, llama90b_df, mixtral_df]
)
```
